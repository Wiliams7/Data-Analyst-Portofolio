```python
# ==========================================
#  Import Library
# ==========================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


print("Semua library berhasil di-import!")
```

```python
# ==========================================
# Load Dataset
# ==========================================
import pandas as pd

df = pd.read_csv(
    r"C:\File-file Aplikasi\Data Analyst\Database\Excel\Sample - Superstore.csv",
    encoding="latin1"
)

df.head()
```

```python
# ==========================================
# Data exploration
# ==========================================
df.shape
df.info()
df.describe()
```

```python
# ==========================================
# DATA QUALITY CHECK
# ==========================================

# 1. missing values
print("MISSING VALUES")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())


# 2. duplicate rows
print("\nDUPLICATE DATA")
print("Total Duplicate Rows:", df.duplicated().sum())


# 3. tYpe data
print("\nDATA TYPES")
print(df.dtypes)


# 4. Checking number of rows and columns
print("\nDATA DIMENSION")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# ==========================================
# DATA CLEANING
# ==========================================

# 1. Check type data and date
print("DATE COLUMN TYPES")
print(df[["Order Date", "Ship Date"]].dtypes)


# 2. Mengubah kolom tanggal menjadi datetime
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)


# 3. Checking the date conversion results
print("\nDATE CONVERSION CHECK")
print(df[["Order Date", "Ship Date"]].dtypes)

print("\nMissing Dates After Conversion:")
print(df[["Order Date", "Ship Date"]].isnull().sum())


# 4. Checking the date range
print("\nDATE RANGE")
print("Earliest Order:", df["Order Date"].min())
print("Latest Order:", df["Order Date"].max())


# 5. Checking numerical data
print("\nNUMERIC DATA SUMMARY")
print(df[["Sales", "Quantity", "Discount", "Profit"]].describe())
```

```python
# ==========================================
# SALES PERFORMANCE ANALYSIS
# ==========================================

# 1. Key Performance Indicators (KPI)
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()

profit_margin = (total_profit / total_sales) * 100

print("===== BUSINESS KPI =====")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity: {total_quantity:,}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Profit Margin: {profit_margin:.2f}%")


# 2. Sales and Profit by Category
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).sort_values(by="Total_Sales", ascending=False)

print("\n===== SALES BY CATEGORY =====")
display(category_analysis)


# 3. Sales and Profit by Region
region_analysis = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).sort_values(by="Total_Sales", ascending=False)

print("\n===== SALES BY REGION =====")
display(region_analysis)


# 4. Sales and Profit by Segment
segment_analysis = df.groupby("Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).sort_values(by="Total_Sales", ascending=False)

print("\n===== SALES BY SEGMENT =====")
display(segment_analysis)


# 5. Sales and Profit by Ship Mode
ship_mode_analysis = df.groupby("Ship Mode").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).sort_values(by="Total_Sales", ascending=False)

print("\n===== SALES BY SHIP MODE =====")
display(ship_mode_analysis)
```

```python

# ==========================================
# PRODUCT ANALYSIS
# ==========================================

# 1. Sales and Profit by Sub-Category
subcategory_analysis = df.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).sort_values(by="Total_Sales", ascending=False)

print("===== SALES BY SUB-CATEGORY =====")
display(subcategory_analysis)


# 2. Top 10 Products by Sales
top_products = df.groupby("Product Name").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).sort_values(by="Total_Sales", ascending=False).head(10)

print("\n===== TOP 10 PRODUCTS BY SALES =====")
display(top_products)


# 3. Top 10 Loss-Making Products
loss_products = df.groupby("Product Name").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).query("Total_Profit < 0").sort_values(
    by="Total_Profit", ascending=True
).head(10)

print("\n===== TOP 10 LOSS-MAKING PRODUCTS =====")
display(loss_products)


# ==========================================
# TIME SERIES ANALYSIS
# ==========================================

# 4. Sales and Profit by Year
df["Order Year"] = df["Order Date"].dt.year

yearly_analysis = df.groupby("Order Year").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

print("\n===== YEARLY SALES ANALYSIS =====")
display(yearly_analysis)


# 5. Monthly Sales Trend
df["Order Month"] = df["Order Date"].dt.to_period("M")

monthly_analysis = df.groupby("Order Month").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

print("\n===== MONTHLY SALES TREND =====")
display(monthly_analysis)
```

```python

# ==========================================
# DATA VISUALIZATION
# ==========================================
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Sales by Category
plt.figure(figsize=(8, 5))
sns.barplot(
    data=category_analysis.reset_index(),
    x="Category",
    y="Total_Sales"
)
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()


# 2. Profit by Region
plt.figure(figsize=(8, 5))
sns.barplot(
    data=region_analysis.reset_index(),
    x="Region",
    y="Total_Profit"
)
plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()


# 3. Sales by Segment
plt.figure(figsize=(8, 5))
sns.barplot(
    data=segment_analysis.reset_index(),
    x="Segment",
    y="Total_Sales"
)
plt.title("Total Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()


# 4. Top 10 Products by Sales
plt.figure(figsize=(10, 6))
sns.barplot(
    data=top_products.reset_index(),
    x="Total_Sales",
    y="Product Name"
)
plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()


# 5. Monthly Sales Trend
monthly_chart = monthly_analysis.reset_index()
monthly_chart["Order Month"] = monthly_chart["Order Month"].astype(str)

plt.figure(figsize=(14, 5))
sns.lineplot(
    data=monthly_chart,
    x="Order Month",
    y="Total_Sales",
    marker="o"
)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 6. Monthly Profit Trend
plt.figure(figsize=(14, 5))
sns.lineplot(
    data=monthly_chart,
    x="Order Month",
    y="Total_Profit",
    marker="o"
)
plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# ==========================================
# BUSINESS INSIGHTS
# ==========================================

# 1. Highest Sales Category
best_category = category_analysis["Total_Sales"].idxmax()
best_category_sales = category_analysis["Total_Sales"].max()

# 2. Highest Profit Region
best_region = region_analysis["Total_Profit"].idxmax()
best_region_profit = region_analysis["Total_Profit"].max()

# 3. Highest Sales Segment
best_segment = segment_analysis["Total_Sales"].idxmax()
best_segment_sales = segment_analysis["Total_Sales"].max()

# 4. Best-Selling Product
best_product = top_products["Total_Sales"].idxmax()
best_product_sales = top_products["Total_Sales"].max()

# 5. Most Profitable Sub-Category
best_subcategory = subcategory_analysis["Total_Profit"].idxmax()
best_subcategory_profit = subcategory_analysis["Total_Profit"].max()

# 6. Least Profitable Sub-Category
worst_subcategory = subcategory_analysis["Total_Profit"].idxmin()
worst_subcategory_profit = subcategory_analysis["Total_Profit"].min()

# 7. Top 10 Loss-Making Products
loss_products = df.groupby("Product Name").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).query("Total_Profit < 0").sort_values(
    by="Total_Profit", ascending=True
).head(10)

# 8. Lowest Sales Category
worst_category = category_analysis["Total_Sales"].idxmin()
worst_category_sales = category_analysis["Total_Sales"].min()


# ==========================================
# DISPLAY BUSINESS INSIGHTS
# ==========================================

print("=" * 50)
print("             BUSINESS INSIGHTS")
print("=" * 50)

print(f"\n1. HIGHEST SALES CATEGORY")
print(f"   Category: {best_category}")
print(f"   Sales: ${best_category_sales:,.2f}")

print(f"\n2. HIGHEST PROFIT REGION")
print(f"   Region: {best_region}")
print(f"   Profit: ${best_region_profit:,.2f}")

print(f"\n3. HIGHEST SALES SEGMENT")
print(f"   Segment: {best_segment}")
print(f"   Sales: ${best_segment_sales:,.2f}")

print(f"\n4. BEST-SELLING PRODUCT")
print(f"   Product: {best_product}")
print(f"   Sales: ${best_product_sales:,.2f}")

print(f"\n5. MOST PROFITABLE SUB-CATEGORY")
print(f"   Sub-Category: {best_subcategory}")
print(f"   Profit: ${best_subcategory_profit:,.2f}")

print(f"\n6. LEAST PROFITABLE SUB-CATEGORY")
print(f"   Sub-Category: {worst_subcategory}")
print(f"   Profit: ${worst_subcategory_profit:,.2f}")

print(f"\n7. TOP 10 LOSS-MAKING PRODUCTS")
display(loss_products)

print(f"\n8. LOWEST SALES CATEGORY")
print(f"   Category: {worst_category}")
print(f"   Sales: ${worst_category_sales:,.2f}")

print("\n" + "=" * 50)
```
